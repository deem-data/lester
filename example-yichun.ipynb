```python
import pandas as pd
from lester.messy.ecom_task1 import __dataprep__original, __dataprep__manual

products_pathes = [f"data/products-data-{partition}.tsv" for partition in range(0, 3)]
reviews_pathes = [f"data/reviews-{partition}.tsv" for partition in range(0, 3)]

dataprep__original_result = __dataprep__original(products_pathes, reviews_pathes)
dataprep__manual_result  = __dataprep__manual(products_pathes, reviews_pathes)

dataprep_diff = dataprep__original_result.compare(dataprep__manual_result.df[dataprep__original_result.columns])
assert len(dataprep_diff) == 0

```

```python
from lester.messy.ecom_task1 import __datasplit__original, __datasplit__manual

train_original, val_original, test_original = __datasplit__original(dataprep__original_result)
train_manual, val_manual, test_manual = __datasplit__manual(dataprep__manual_result)

assert len(train_original) == len(train_manual.df)
assert len(val_original) == len(val_manual.df)
assert len(test_original) == len(test_manual.df)
```
