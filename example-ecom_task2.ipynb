```python
import pandas as pd
import lester as ld
import numpy as np

from lester.messy.ecom_task2 import __datasplit__original, __datasplit__manual

products = pd.read_csv(f"data/products-data-0.tsv", sep='\t', names=['product_id', 'product_category', 'product_name'])
products_tracked = ld.read_csv(f"data/products-data-0.tsv", sep='\t', names=['product_id', 'product_category', 'product_name'], header=None)

reviews = pd.read_csv(f"data/reviews-0.tsv", sep='\t', names=['product_id', 'rating', 'review'])
reviews_tracked = ld.read_csv(f"data/reviews-0.tsv", sep='\t', names=['product_id', 'rating', 'review'], header=None)

products_with_reviews = products.merge(reviews, on='product_id')
products_with_reviews_tracked = products_tracked.join(reviews_tracked, left_on='product_id', right_on='product_id')

train_original, test_original = __datasplit__original(products_with_reviews)
train_manual, test_manual = __datasplit__manual(products_with_reviews_tracked)

assert len(train_original) == len(train_manual.df)
assert len(test_original) == len(test_manual.df)

from lester.messy.ecom_task2 import __featurise_original, __featurise_manual

features_original = __featurise_original(products_with_reviews)
features_manual = __featurise_manual(products_with_reviews_tracked.df)

assert features_original.shape == features_manual.shape
assert np.allclose(features_original[:,0], features_manual[:,0])
# TODO how should we compare the low dimensional projections?
```
